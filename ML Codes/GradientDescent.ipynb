# ML Lab Practical 5: Gradient Descent Algorithm
**Objective:** Find the minimum of \(f(x)=(x+3)^2\) using gradient descent, starting at \(x=2\).

**Learning rate:** 0.1 | **Iterations:** 100

```python
# Define the function
def f(x):
    return (x + 3) ** 2

# Define the derivative
def derivative(x):
    return 2 * (x + 3)
```

```python
# Initial values
x = 2
learning_rate = 0.1
iterations = 100
```

```python
# Gradient Descent
for i in range(iterations):
    gradient = derivative(x)
    x = x - learning_rate * gradient
```

```python
# Display the result
print("Minimum point x =", round(x, 6))
print("Minimum value y =", round(f(x), 10))
```
